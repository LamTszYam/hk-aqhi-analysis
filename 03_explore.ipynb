```python
import pandas as pd
data = pd.read_csv("data/aqhi_2026_h1_clean.csv", parse_dates=["Date"])
print(data.drop(columns=["Date", "Hour"]).describe().T.sort_values("mean"))
```

```python
data[data["Tung Chung"] == 11][["Date", "Hour", "Tung Chung", "Tuen Mun", "Yuen Long"]]
```

```python
import matplotlib.pyplot as plt

day = data[data["Date"] == "2026-01-13"]
day.plot(x="Hour", y=["Tung Chung", "Tuen Mun", "Yuen Long", "Central", "Mong Kok"])
plt.title("AQHI on 13 Jan 2026")
plt.ylabel("AQHI")
plt.show()
```
