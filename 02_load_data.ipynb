```python
import pandas as pd
df = pd.read_csv("data/202606_Eng.csv", skiprows=7, na_values=["-", "-*"])
df.head()
```

```python
print(df.shape)
df.info()
```

```python
df["Date"] = df["Date"].ffill()
df.head()
df = df[df["Hour"] != "Daily Max"]
print(df.shape)
```

```python
stations = df.columns[2:]
for col in stations:
    df[col] = df[col].str.replace("*", "", regex=False)
    df[col] = pd.to_numeric(df[col])
df.info()
```
